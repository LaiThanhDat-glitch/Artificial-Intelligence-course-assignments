```python
def TongChanLe(list):
    TC = TL = 0
    for element in list:
        if element % 2 == 0:
            TC += element
        else:
            TL += element
    return TC, TL

mylist = [1, 2, 3, 4, 5, 6, 7, 8, 9]
tc, tl = TongChanLe(mylist)
print("Tong chan: " + str(tc) + ", Tong le: " + str(tl))
```

```python
from math import pi, sin

def f(x):
    return x * sin(x)

a = []
step = pi/200
for i in range(101):
    a.append(i*step)

ff = []
for i in range(101):
    ff.append(f(a[i]))

TP = 0
for i in range(101):
    TP += ff[i]
TP *= step
print(TP)
```

```python
class Fraction(object):
    __nominator = 0
    __denominator = 1

    def __init__(self, n, d):
        if d == 0:
            return Exception()
        self.__nominator = n
        self.__denominator = d

    def __str__(self):
        n = self.__nominator
        d = self.__denominator
        while n != d:
            if n > d:
                n = n - d
            else:
                d = d - n
        self.__nominator //= n
        self.__denominator //= d
        return str(self.__nominator) + "/" + str(self.__denominator)

    def __add__(self, other):
        n = self.__nominator * other.__denominator + self.__denominator * other.__nominator
        d = self.__denominator * other.__denominator

ps1 = Fraction(4, 2)
ps2 = Fraction(2, 3)
print(ps1 + ps2)
```
